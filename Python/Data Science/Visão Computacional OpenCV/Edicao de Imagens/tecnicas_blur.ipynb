```python
import cv2
import numpy as np
```

```python

def show_img(titulo, image):
    cv2.imshow(titulo, image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
img = cv2.imread('assets/fotos/cat.jpg')
show_img("Imagem Padrão", img)
```

```python
# Blur por média
average = cv2.blur(img, (6,6))
show_img('Blur por média', average) 
```

```python
# Gaussian Blur

gauss = cv2.GaussianBlur(img,(7,7), 0)
show_img('Gaussian Blur', gauss)
```

```python
# Blur por Mediana

median = cv2.medianBlur(img, 9)
show_img('Median Blur', median)
```

```python
# Filtro Bilateral - Filtro de suavização não linear, com preservações de bordas e redução de ruído para imagens. 
# Substitui a intensidade de cada pixel por uma média ponderada de valores de intensidade dos pixels próximos. Esse efeito reduz o ruído.
# sigmaColor -> Filtra o Sigma no ColorSpace
# sigmaSpace -> Filtra o Sigma no espaço de coordenadas.
# Bilateral Filter (imagem, d, sigamColor, sigmaSpace)
# d = kernel size (diametro de analise do modelo)
img_noisy  = cv2.imread('assets/fotos/noisy_image.jpg')
show_img('Imagem com ruido', img_noisy)

bilateral = cv2.bilateralFilter(img_noisy, 15, 75, 75)
show_img('Filtro Bilateral', bilateral)
```
