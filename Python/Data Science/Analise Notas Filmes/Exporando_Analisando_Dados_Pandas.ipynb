IMPORTANDO DADOS DOS FILMES - CATEGORIAS:

```python
filmes = pd.read_csv ("C:\\Users\\RedSeek\\Desktop\\DATA SCIENCE FORMACAO\\Explorando e Analisando Dados\\ml-latest-small\\movies.csv")
```

```python
filmes.columns = ["filme_ID","titulo","generos"]
```

RETORNANDO APENAS AS NOTAS DO FILME TOY STORY - FAZENDO PESQUISA NO DATAFRAME:

```python
notas.query("filmeId==1")
```

RETORNANDO APENAS AS NOTAS DO FILME TOY STORY - FAZENDO PESQUISA NO DATAFRAME - SOMENTE A MEDIA:

```python
notas.query("filmeId==1")["nota"].mean()
```

RETORNANDO APENAS AS NOTAS DO FILME JUMANJI - FAZENDO PESQUISA NO DATAFRAME - SOMENTE A MEDIA:

```python
notas.query("filmeId==2")["nota"].mean()
```

CALCULANDO A MEDIA DE TODOS OS FILMES POR FILME - AGRUPANDO AS NOTAS POR FILME:

```python
medias_por_filme = notas.groupby("filmeId")["nota"].mean()
```

```python
medias_por_filme.head()
```

VISUALIZANDO OS DADOS NO GRAFICO DE DISTRUICAO HISTOGRAMA - MEDIAS POR FILME - AVALIACAO:

```python
medias_por_filme.plot(kind="hist")
```

VISUALIZANDO A MEDIANA DOS FILMES - GRAFICO DE DISTRIBUIÇAO - DISPLOT HISTOGRAMA NO SEABORN E CUSTOMIZANDO GRAFICO COM MATPLOTLIB:

```python
import matplotlib.pyplot as plt
sns.displot(medias_por_filme, kde=True)
plt.title("Histograma das medias dos filmes")
```
