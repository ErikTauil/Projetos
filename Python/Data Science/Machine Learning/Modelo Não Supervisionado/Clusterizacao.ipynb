# Aplicando o valor de inércia

A inércia é uma medida que nos permite avaliar a qualidade dos clusters, pois retorna a soma das distâncias ao quadrado entre os pontos dentro do cluster e seu centroide.

Quando temos valores mais altos de inércia, os pontos estão mais distantes do centroide, no centro do cluster. Quanto menor a inércia, mais próximos estão esses valores e melhor estará o agrupamento, pois os dados estarão mais próximos do centro — ou seja, estão mais agrupados. O ideal é ter o valor mais baixo possível de inércia.

```python
mod_kmeans.inertia_
```

# Aplicando o valor de silhueta

A silhueta retorna o valor para a separação e coesão dos pontos dentro de um cluster, comparando-os em relação aos pontos próximos dentro de si e de outro cluster próximo.

Esses valores de silhueta são dados para cada ponto ou amostra que o K-Means agrupou dentro de um cluster. Os valores variam entre 1 e -1, sendo 1 um valor ótimo, indicando que o ponto está bem agrupado dentro do cluster.

Quanto mais próximo de zero, mais próximo o ponto estará das fronteiras do cluster. Quando o valor é negativo, próximo a -1, significa que o ponto provavelmente está agrupado de forma errada, num cluster onde não deveria estar.

```python
from sklearn.metrics import silhouette_score

silhouette_score(dados, mod_kmeans.predict(dados))
```

Esse valor está mais próximo de zero do que de 1. Ou seja, os valores de silhueta para cada ponto indicam que, em média, os dados estão mais próximos da fronteira de outro cluster do que do centroide de seu próprio cluster.

Com essas duas informações, o valor de inércia e o valor de silhueta, podemos concluir que, provavelmente, dois clusters não trazem a quantidade certa de agrupamentos ou os agrupamentos de melhor qualidade para dados. Com esse valor, não teremos sucesso para solucionar o desafio.

# Comparando valores de K-Means

valores de métricas que coletado do modelo criado com o K-Means, não obtive o melhor resultado para os agrupamentos em termos de qualidade. É necessário verificar se foi feito uma má escolha na inicialização dos clusters e na quantidade de clusters que poderíamos utilizar com o K-Means.

Para avaliar esse ponto, será utilizado as mesmas métricas de avaliação do modelo, a fim de avaliar modelos K-Means para diferentes quantidades de clusters, como 3, 4, 5, e assim por diante.

Será Avaliado esses mesmos valores e a silhueta para diferentes inicializações de clusters. Assim, teremos a garantia de que foi feita uma boa escolha com um bom valor de cluster inicial, se há outra quantidade de clusters mais interessante, ou se o problema é outro.

```python
def avaliacao(dados):
    inercia = []
    silhueta = []

    for k in range(2,21):
        kmeans = KMeans(n_clusters=k, random_state=45, n_init='auto')
        kmeans.fit(dados)
        inercia.append(kmeans.inertia_)
        silhueta.append(f'k={k} - '+ str(silhouette_score(dados, kmeans.predict(dados))))
        
    return silhueta, inercia
```

```python
silhueta, inercia = avaliacao(dados)
silhueta
```

Podemos observar que o maior e melhor valor de silhueta é para k=2, ou seja, para a quantidade de 2 clusters construídos pelo K-Means. Isso indica que, independentemente dos valores de cluster que escolhermos para esse modelo, o resultado não melhorará muito além do que já analisamos.

# Criando um gráfico de distribuição de silhuetas

Iniciamos a avaliação de desempenho do modelo K-Means para diferentes valores de cluster a partir do valor médio de silhuetas. O problema é que, com apenas o valor médio, temos apenas uma visão geral de como está a distribuição do valor de silhueta para cada ponto de amostra. Não conseguimos entender como se deu essa distribuição.

Para obter uma visão mais detalhada e resolver essa questão, podemos criar um gráfico que permita analisar a distribuição de valores de silhueta por amostra em diferentes clusters. 

```python
import matplotlib.cm as cm
from sklearn.metrics import silhouette_samples

def graf_silhueta (n_clusters, dados):

    # Aplica o KMeans ao conjunto de dados
    kmeans = KMeans(n_clusters=n_clusters, random_state=45, n_init = 'auto')
    cluster_previsoes = kmeans.fit_predict(dados)
    
    # Calcula o silhouette score médio
    silhueta_media = silhouette_score(dados, cluster_previsoes)
    print(f'Valor médio para {n_clusters} clusters: {silhueta_media:.3f}')
    
    # Calcula a pontuação de silhueta para cada amostra
    silhueta_amostra = silhouette_samples(dados, cluster_previsoes)
    
    # Configuração da figura para o gráfico de silhueta
    fig, ax1 = plt.subplots(1, 1)
    fig.set_size_inches(9, 7)
    
    # Limites do gráfico de silhueta
    ax1.set_xlim([-0.1, 1])
    ax1.set_ylim([0, len(dados) + (n_clusters + 1) * 10])
    
    y_lower = 10
    for i in range(n_clusters):
        ith_cluster_silhueta_amostra = silhueta_amostra[cluster_previsoes == i]
        ith_cluster_silhueta_amostra.sort()
        
        tamanho_cluster_i = ith_cluster_silhueta_amostra.shape[0]
        y_upper = y_lower + tamanho_cluster_i
        
        cor = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper), 0, ith_cluster_silhueta_amostra, 
            facecolor=cor, edgecolor=cor, alpha=0.7)

        ax1.text(-0.05, y_lower + 0.5 * tamanho_cluster_i, str(i))
        y_lower = y_upper + 10  # 10 para o espaço entre gráficos
        
    # Linha vertical para a média do Silhouette Score
    ax1.axvline(x=silhueta_media, color='red', linestyle='--')
        
    ax1.set_title(f'Gráfico da Silhueta para {n_clusters} clusters')
    ax1.set_xlabel('Valores do coeficiente de silhueta')
    ax1.set_ylabel('Rótulo do cluster')
    
    ax1.set_yticks([])  # Remove os ticks do eixo y
    ax1.set_xticks([i/10.0 for i in range(-1, 11)])
    
    plt.show()
```

Vamos verificar o que está acontecendo dentro desse código para executar a célula e verificar como se deu a distribuição dos pontos dos clusters.

A primeira coisa que acontece no código é a importação do matplotlib para visualizar o resultado e da métrica silhouette_samples() para obter os valores de silhueta por amostras. Afinal, não verificaremos somente a média.

O nome da função que gera esse gráfico é graf_silhueta(). Ela recebe entre parênteses o valor de clusters que achamos ser o melhor para avaliar e os nossos dados.

A partir disso, dentro da função, criamos o modelo kmeans com a quantidade específica de clusters, o random_state 45, e o init igual a auto para evitar o aviso. A partir daí, treinamos o modelo e prevemos os dados com o fit_predict().

Em seguida, temos um código para visualizar o valor médio de silhueta, para realizar uma comparação entre o texto e o gráfico, analisando o ponto exato do valor médio. Fazemos isso com o silhouette_score(), utilizado anteriormente, e depois, imprimimos o valor médio.

Mais abaixo, coletamos os valores de silhueta por amostra com o silhouette_samples(). Em seguida, começamos a configurar a visualização com o Matplotlib, criando a figura e determinando os axis.

Mais abaixo, entramos em um for para verificar por clusters e ajustar as cores do gráfico, possibilitando visualizar nitidamente a diferença da distribuição de silhuetas entre as quantidades de clusters 0 e 1, criadas com a função KMeans().

Depois de configurar as cores e a forma de visualização, finalizamos a criação do plot, colocando uma linha vertical exatamente no ponto onde temos o valor de silhueta médio. Além disso, configuramos os títulos, labels, e outro detalhes do modelo.

```python
graf_silhueta(2, dados)  # quantidade de clusters = 2
```

Acima do gráfico, já temos o resultado inicial de 0,38, que é o valor de cluster médio. No gráfico, podemos ver os dois clusters, 0 e 1, e uma linha tracejada com o valor médio de cluster.

Observamos que, para o Cluster 1, representado pela extensa área verde, temos inicialmente uma boa quantidade de pontos em seu interior com valores positivos, mais próximos de 1, como o valor de silhueta. No entanto, ao avaliar outras amostras, vemos uma dispersão dos resultados, com vários pontos de valor de silhueta próximos a zero.

Infelizmente, percebemos que esse agrupamento não tem a melhor qualidade, mesmo apresentando bons resultados no início. Conforme vemos outras amostras, constatamos que elas se dispersam muito.

Outro cluster a analisar é o 0, que teve um resultado pior que o 1. Seus valores, como um todo, são muito próximos a zero. Inclusive, temos valores negativos, indicando que vários pontos, além de estarem distantes do centro e próximos das fronteiras, podem estar agrupados erroneamente dentro desse cluster.

Esses pontos analisados mostram que a qualidade desse conjunto está ruim.

Conclusão
Com isso, percebemos de maneira aprofundada que o melhor resultado ainda não é bom. Isso pode indicar que o problema não está na inicialização de clusters que definimos, mas sim numa questão que seria resolvida ajustando os dados.

# Analisando o modelo

Ao avaliar os valores de silhueta, percebemos que nosso modelo K-Means não melhorará com a alteração dos valores de cluster na inicialização. No entanto, ainda precisamos garantir que esse pensamento é válido.

Para isso, podemos analisar a métrica de inércia. Normalmente, utilizamos o método cotovelo para essa avaliação. 

O método cotovelo é uma técnica visual que nos permite avaliar e identificar a quantidade ideal de clusters que proporciona o melhor valor de inércia, melhorando a qualidade dos agrupamentos.

Esse método consiste em plotar os valores de inércia em relação à quantidade de clusters. Escolhemos a quantidade de clusters a partir do ponto em que o valor de inércia para de decair significativamente e começa a cair lentamente. No ponto onde ocorre essa quebra na curva, observamos um "cotovelo", que seria o ponto K ideal, ou seja, a quantidade de clusters ideal.

```python
# função lot_cotovelo(). Entre os parênteses, a única informação necessária são os valores de inércia, que já foi coletado anteriormente com a função de avaliação. 
# Portanto, foi definido inercia como os dados que precisa ser recebido.

# no eixo X, quantidade de clusters, variando de 2 a 20, utilizando range(2, 21). 
# No eixo Y, valores de inercia. 
# Para a visualização, foi utilizado o código 'bo-', onde b define a cor azul, o marca os pontos de correlação entre as duas variáveis, e - define a linha como reta.

def plot_cotovelo(inercia):
    plt.figure(figsize=(8,4))
    plt.plot(range(2,21),inercia,'bo-')
    plt.xlabel('Número de clusters')
    plt.ylabel('Inércia')
    plt.title('Método do Cotovelo para Determinação de k')
    plt.show()
```

```python
plot_cotovelo(inercia)
```

O gráfico em azul mostra os pontos bem marcados. No topo do eixo y, observamos o valor 1e7, informando que todos os valores foram multiplicados por 10⁷. Isso indica que a inércia é muito alta.

Conclusão
Embora os valores decaiam com o aumento da quantidade de clusters conforme esperado, ainda são elevados. Não encontramos um ponto de quebra representando o "cotovelo" — isso indica que, independentemente dos valores de cluster, o desempenho do modelo não melhora.
