RESETANDO O INDEX PARA CRIAR UM NOVO INDEX NUMERICO:

```python
contagem_de_lingua = tmdb["original_language"].value_counts().to_frame().reset_index()
```

RENOMEANDO AS COLUNAS DO DATAFRAME CRIADO:

```python
contagem_de_lingua.columns = ["original_language", "total"]
```

```python
contagem_de_lingua.head()
```

PLOTANDO - VISUALIZANDO OS DADOS DO NOVO DATAFRAME CONTAGEM DE LINGUA NO GRAFICO DE BARRAS - BARPLOT SEABORN:

```python
sns.barplot(data = contagem_de_lingua, x= "original_language", y="total")
```

PLOTANDO - VISUALIZANDO OS DADOS DO NOVO DATAFRAME CONTAGEM DE LINGUA NO GRAFICO DE BARRAS - COUNTPLOT SEABORN:

```python
sns.countplot(data=tmdb, x="original_language")
```

EXPLORANDO OS VALORES TOTAIS DE FILMES POR LINGUA:

```python
total_por_lingua = tmdb["original_language"].value_counts()
total_geral = total_por_lingua.sum()
total_de_ingles = total_por_lingua.loc["en"]
total_do_resto = total_geral - total_de_ingles
print (total_geral,total_de_ingles,total_do_resto) 
```

CRIANDO UM DICIONARIO PYTHON COM OS DADOS TOTAIS GERADOS:

```python
dados = {
  "lingua" : ["ingles", "outros"],
  "total"  : [total_de_ingles, total_do_resto]
}
```

TRANSFORMANDO O DICIONARIO EM UM DATAFRAME COM OS DADOS TOTAIS GERADOS:

```python
dados = pd.DataFrame(dados)
dados
```

PLOTANDO VISUALIZANDO COM O GRAFICO DE LINHA - TOTAL DE LINGUA INGLES X TOTAL DAS DEMAIS LINGUAS EM GERAL - SEABORN:

COMO FOI IDENTIFICADO ANTERIORMENTE QUE A LINGUA INGLESA TEM O MAIOR VALOR, FOI FEITO UM COMPARATIVO COM AS DEMAIS LINGUAS NO GRAFICO DE BARRAS ABAIXO. 

```python
sns.barplot(data=dados, x="lingua", y="total")
```

PLOTANDO VISUALIZANDO COM O GRAFICO PIZZA - TOTAL DE LINGUA INGLES X TOTAL DAS DEMAIS LINGUAS EM GERAL:

#A Melhor visualizaçao neste caso do comparativo dos valores totais e o Grafico de Barras Acima em relacao ao grafico de pizza abaixo.

```python
dados.plot(kind="pie", y="total", labels=dados["lingua"])
```

EXPLORANODO O RESTO / DEMAIS LINGUAS

Filmes que nao sao em Ingles:

```python
total_de_outros_filmes_por_lingua = tmdb.query("original_language != 'en'")["original_language"].value_counts()
```

```python
total_de_outros_filmes_por_lingua
```

```python
total_de_outros_filmes_por_lingua.head() 
```

PLOTANDO / VISUALIZANDO A CONTAGEM DOS FILMES QUE NAO SAO INGLES - SEABORN:

```python
sns.countplot(data=tmdb.query("original_language !='en'"), 
              x="original_language")
```

ORDENANDO AS BARRAS DO GRAFICO DO MAIOR VALOR PARA O MENOR VALOR:

```python
sns.countplot(data=tmdb.query("original_language !='en'"), 
              order=total_de_outros_filmes_por_lingua.index,
              x="original_language")
```

AJUSTANDO AS CORES DAS BARRAS DE ACORDO COM A ORDEM DAS BARRAS DO MAIO PARA O MENO VALOR:

```python
sns.countplot(data=tmdb.query("original_language !='en'"), 
              order=total_de_outros_filmes_por_lingua.index,
              palette="mako",
              hue="original_language",
              hue_order=total_de_outros_filmes_por_lingua.index,
              x="original_language")
plt.title("Distribuiçao da lingua original nos filmes exceto em ingles")
plt.show()
```

EXPLORANDO O PERCENTUAL DO TOTAL DE CADA LINGUA, EXCETO O INGLES:

```python
tmdb.query("original_language != 'en'")["original_language"].value_counts(normalize=True)
```

PLOTANDO O MESMO GRAFICO, POREM VISUALIZANDO PELOS VALORES EM PERCENTUAIS:

```python
sns.countplot(data=tmdb.query("original_language !='en'"), 
              order=total_de_outros_filmes_por_lingua.index,
              palette="mako",
              hue="original_language",
              hue_order=total_de_outros_filmes_por_lingua.index,
              stat="percent",
              x="original_language")
plt.title("Distribuiçao da lingua original nos filmes exceto em ingles")
plt.show()
```

CONFIGURANDO O TAMANHO DO PLOT/GRAFICO - MELHORANDO A VISUALIZACAO DAS LEGENDAS:

```python
plt.figure(figsize=(16,8))
sns.countplot(data=tmdb.query("original_language !='en'"), 
              order=total_de_outros_filmes_por_lingua.index,
              palette="mako",
              hue="original_language",
              hue_order=total_de_outros_filmes_por_lingua.index,
              stat="percent",
              x="original_language")
plt.title("Distribuiçao da lingua original nos filmes exceto em ingles")
plt.show()
```
