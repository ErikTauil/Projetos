```python
import pandas as pd
df = pd.read_csv ("C:\\Users\\RedSeek\\Desktop\\DATA SCIENCE FORMACAO\\Imigrantes do Brasil para o Canada\\imigrantes_canada.csv")
df
```

ANALISANDO E EXPLORANDO OS DADOS - VERIFICANDO SE POSSUI ALGUM DADO NULO PARA SER TRATADO:

```python
df.info()
```

ANALISAR AS TENDENCIAS DE IMIGRACAO DO BRASIL EM UM DETERMINADO PERIODO:

ALTERANDO O INDEX DO DATAFRAME - SETANDO A COLUNA País COMO INDEX:

```python
df.set_index('País', inplace=True)
```

CRIANDO A VARIAVEL PARA ARMAZENAR EM UMA LISTA O INTERVALO DE TEMPO DOS ANOS:

```python
anos = list(map(str,range(1980, 2014))) 
```

```python
anos
```

BUSCANDO SOMENTE OS DADOS REFERENTE AO BRASIL DENTRO DO INTERVALO DEFINIDO NA VARIAVEL "ANOS":

```python
brasil = df.loc['Brasil', anos]
```

```python
brasil
```

CONVERTENDO A SERIE CRIADA EM UM DATAFRAME:

1 - CRIANDO O DICIONARIO E CONVERTENDO A SERIE EM UMA LISTA:

```python
brasil_dict = {'ano' : brasil.index.tolist(), 'imigrantes' : brasil.values.tolist()}
```

2 - CRIANDO O DATAFRAME:

```python
dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil
```

CRIANDO O GRAFICO DE LINHAS (COM MATPLOTLIB) PARA ENTENDER AS TENDENCIAS DE IMIGRACAO DO BRASIL:

```python
import matplotlib.pyplot as plt
```

AJUSTANDO O GRAFICO, TAMANHO, INTERVALO DOS EIXOS X e Y, CRIANDO TITULO E ROTULOS - E PLOTANDO O GRAFICO:

```python
plt.figure(figsize=(8,4)) #ajustando o Tamanho do Grafico
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigraçao do Brasil para o Canada')
plt.xlabel('Ano')
plt.ylabel('Numero de Imigrantes')
plt.xticks(['1980','1985','1990','1995','2000','2005', '2010']) #ajustando o intervalo de anos do eixo X
plt.yticks([500, 1000, 1500, 2000, 2500, 3000]) #ajustando o intervalo de valores do eixo y
plt.show()
```

CRIANDO FIGURA EM PYTHON - CUSTOMIZANDO O GRAFICO DE OUTRA FORMA:

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3, color='g')# lw definindo a expessura 3 em pixels da linha do Grafico e inserindo a cor da linha como verde
#marker='o' inserindo marcadores em circulos na linha do Grafico
ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', fontsize=18, loc='left') #criando o Titulo a Esquerda
ax.set_xlabel('Ano', fontsize=14) #criando o rotulo do eixo x
ax.set_ylabel('Numero de Imigrantes',fontsize=14) #criando o rotulo do eixo y
ax.xaxis.set_tick_params(labelsize=12) #ajustando o tamanho dos ticks valores do eixo x
ax.yaxis.set_tick_params(labelsize=12) #ajustando o tamanho dos ticks valores do eixo y
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #criando uma frequencia de 5 e 5 anos dentro do intervalo de Anos
#plt.grid(linestyle= '--') #inserindo a Grade no fundo do Grafico
plt.show()
```

CRIANDO SUBPLOTS EM UMA DIREÇAO NA FIGURA PARA CONTER MAIS DE UM GRAFICO NA FIGURA:

```python
fig, axs = plt.subplots(1,2, figsize=(15,5)) # definido uma linha e 2 subplots e o tamanho

#posiçao 0 referente a criaçao do Primeiro Grafico - GRAFICO DE LINHAS
axs[0].plot(dados_brasil['ano'], dados_brasil['imigrantes'])
axs[0].set_title('Imigração do Brasil para o Canadá\n1980 a 2013')
axs[0].set_xlabel('Ano')
axs[0].set_ylabel('Número de imigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5)) # criando uma frequencia de 5 e 5 anos dentro do intervalo de Anos
axs[0].grid() #adicionando a grade no fundo do Grafico

#posiçao 1 referente a criaçao do Segundo Grafico - BOXPLOT
axs[1].boxplot(dados_brasil['imigrantes'])
axs[1].set_title('Boxplot da Imigração do Brasil para o Canadá\n1980 a 2013')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('Número de imigrantes')
axs[1].grid() #adicionando a grade no fundo do Grafico

plt.show()
```

ANALISANDO AS TENDENCIAS DE IMIGRACAO PARA OS PAISES DA AMERICA DO SUL, INCLUINDO O BRASIL:

Criando subplots em duas direções e comparando as Tendencias em uma unica Figura - contendo 4 subplots:

```python
fig, axs = plt.subplots(2,2, figsize=(10,6)) # definido 2 subplots em uma linha e 2 subplots na segunda linha e o tamanho
fig.subplots_adjust(hspace=0.5, wspace=0.3) #Ajustando e Melhorando o Espaçamento entre os Graficos
fig.suptitle('Imigração dos quatro maiores países da América do Sul para o Canadá de 1980 a 2013') #criando um titulo geral para os 4 Graficos

#criando o subplot do Brasil
axs[0,0].plot(df.loc['Brasil', anos])
axs[0,0].set_title('Brasil')

#criando o subplot da Colombia
axs[0,1].plot(df.loc['Colômbia', anos])
axs[0,1].set_title('Colômbia')

#criando o subplot da Argentina
axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title('Argentina')

#criando o subplot do Peru
axs[1,1].plot(df.loc['Peru', anos])
axs[1,1].set_title('Peru')

#ajustando os valores do eixo x de todos os Graficos com o "for",  criando uma frequencia de 5 e 5 anos dentro do intervalo de Anos
for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))


#ajustando os valores do eixo x e y de todos os Graficos com o "for",  criando os Rotulos
for ax in axs.flat:
  ax.set_xlabel('Ano')
  ax.set_ylabel('Número de imigrantes')

ymin = 0
ymax = 7000 #limitando os valores do eixo y de 0 a 7000 para os 4 Graficos da Figura

for ax in axs.ravel():
  ax.set_ylim(ymin, ymax)

plt.show()
```

```python
#No Brasil, temos dois picos maiores: o primeiro entre 1990 e 1995, e depois um grande pico entre 2005 e 2010. 
#Na Colômbia, não há picos no começo, mas há um grande aumento a partir do final dos anos 90. Então, depois de 1995, temos um aumento bem grande, que aconteceu antes do Brasil. 
#Número de imigrantes máximo no Brasil está em 2.500, enquanto na Colômbia está em 6.000 imigrantes. Temos uma expressividade bem maior na Colômbia.
#Na Argentina, temos dois picos maiores: um entre 1990 e 1995, como no Brasil, e o segundo pico acontecendo também um pouco antes do Brasil, começando nos anos 2000. 
#No Peru, temos dois picos, um começando em 1985 e terminando no final dos anos 90, e outro começando também nos anos 2000, com um maior valor em cerca de 2010.
```

```python
america_sul = df.query('Região == "América do Sul"') #Pesquisando e Selecionando apenas regiao da América do Sul
```

```python
#Plotando o Grafico com Numero total de Imigracoes para cada um dos Paises - Grafico de Barras: 
```

```python
cores = ['royalblue', 'orange', 'forestgreen', 'orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal', 'tomato'] #lista de cores criada
fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(america_sul.index, america_sul['Total'], color=cores) #Colunas Index e Total
ax.set_title('Imigração da América do Sul para o Canadá\n1980 a 2013', loc='left', fontsize=18)
ax.set_xlabel('')
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)

plt.show()
```

CRIANDO GRAFICO DE BARRAS INVERTIDO:

```python
cores = ['royalblue', 'orange', 'forestgreen', 'orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal', 'tomato'] #lista de cores criada
fig, ax = plt.subplots(figsize=(12, 5))
ax.barh(america_sul.index, america_sul['Total'], color=cores) #Colunas Index e Total e uso da funçao "barh" para inverter o Grafico de Barras na Horizonal para uma Melhor Visualizaçao
ax.set_title('Imigração da América do Sul para o Canadá\n1980 a 2013', loc='left', fontsize=18)
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
#ordernando o dataframe pelo valor total do maior para o menor valor
america_sul_sorted = america_sul.sort_values('Total', ascending=True)
```

```python
cores = ['royalblue', 'orange', 'forestgreen', 'orchid', 'purple', 'brown', 'slateblue', 'gray', 'olive', 'navy', 'teal', 'tomato'] #lista de cores criada
fig, ax = plt.subplots(figsize=(12, 5))
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores) #Colunas Index e Total e uso da funçao "barh" para inverter o Grafico de Barras na Horizonal para uma Melhor Visualizaçao
ax.set_title('Imigração da América do Sul para o Canadá\n1980 a 2013', loc='left', fontsize=18)
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
#Guiana é o país com o maior número de imigrantes, seguido de Colômbia, Peru e o Brasil, que ficou em quarto lugar. 
#E os países com menores valores, do de maior para o de menor valor, são Uruguai, Bolívia, Paraguai e Suriname.
```

Destacando a cor da barra do Brasil em relação às outras para indficar que o Brasil é o quarto país com mais imigrantes para o Canadá no período de 1980 a 2013":

```python
cores = []
for pais in america_sul_sorted.index:
    if pais == 'Brasil':
        cores.append('green')
    else:
        cores.append('silver')

fig, ax = plt.subplots(figsize=(12, 5))
ax.barh(america_sul_sorted.index, america_sul_sorted['Total'], color=cores)
ax.set_title('América do Sul: Brasil foi o quarto país com mais imigrantes\npara o Canadá no período de 1980 a 2013', loc='left', fontsize=16)
ax.set_xlabel('Número de imigrantes', fontsize=14)
ax.set_ylabel('')
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)


#Inserindo Anotacoes no Grafico - Inserido os Valores Totais de Cada Barra / de cada Pais
for i, v in enumerate(america_sul_sorted['Total']):
    ax.text(v + 20, i, str(v), color='black', fontsize=10, ha='left', va='center')


ax.set_frame_on(False) #Retirando o Frame (a caixa que está em volta das barras)
ax.get_xaxis().set_visible(False) #Retirando o eixo X do Grafico para melhor visualizaçao uma vez que os valores totais foram inserido ao final de cada barra
ax.tick_params(axis='both', which='both', length=0) #Retirando os ticks do eixo Y

plt.show()
```
